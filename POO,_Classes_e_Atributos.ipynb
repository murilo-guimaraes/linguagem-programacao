##**Introdução Teórica**

A Programação Orientada a Objetos (POO) surgiu como uma solução para a crescente complexidade dos softwares procedurais. Segundo Scheffer (2020), o paradigma orientado a objetos permite que o desenvolvedor mapeie elementos do mundo real para o código através de **Classes**. Uma classe funciona como um molde ou "planta baixa", definindo quais características (**Atributos**) e comportamentos (**Métodos**) um objeto possuirá. No Python, a utilização da palavra-chave `class` e do método especial `__init__` são os pilares para a criação de sistemas modulares, onde cada objeto é uma instância viva dessa definição abstrata.

##**Objetivos**

 - **Definir** a estrutura de uma classe utilizando a sintaxe oficial do Python.
 - **Implementar** o método construtor `__init__` para inicialização de estado.
 - **Demonstrar** o processo de instanciação de múltiplos objetos a partir de uma mesma classe.
 - **Validar** a manipulação de atributos e a independência de estado entre instâncias.

---
###**1. Definição da Classe e Construtor**

Uma classe define o tipo do objeto. O método `__init__` é o responsável por atribuir os valores iniciais (estado) no momento em que o objeto nasce.

```python
class Veiculo:
    def __init__(self, marca, modelo, ano):
        # Atributos de Instância
        self.marca = marca
        self.modelo = modelo
        self.ano = ano
        self.ligado = False

# Criando um objeto (Instanciação)
meu_carro = Veiculo("Toyota", "Corolla", 2024)

print(f"Objeto criado: {meu_carro.marca} {meu_carro.modelo} ({meu_carro.ano})")
```

####**O que observar:**
<a>O parâmetro `self` é fundamental: ele garante que os atributos pertençam especificamente àquela instância. Como explicado por Scheffer, a classe `Veiculo` é apenas o conceito; `meu_carro` é o objeto real ocupando espaço na memória.</a>

---
###**2. Métodos: O Que o Objeto Faz**

Métodos são funções definidas dentro da classe que representam as ações que o objeto pode realizar.

```python
class Veiculo:
    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo
        self.ligado = False

    def ligar(self):
        self.ligado = True
        print(f"O {self.modelo} está ligado agora.")

    def exibir_status(self):
        status = "Ligado" if self.ligado else "Desligado"
        print(f"Status do {self.modelo}: {status}")

# Testando comportamentos
carro_viva = Veiculo("Honda", "Civic")
carro_viva.exibir_status()
carro_viva.ligar()
carro_viva.exibir_status()
```

####**O que observar:**
<a>Os métodos alteram o "estado" interno do objeto. Note que ao chamar `.ligar()`, alteramos o atributo `self.ligado`. Isso demonstra o princípio de que o objeto é responsável por gerenciar seus próprios dados.</a>

---
###**3. Múltiplos Objetos, Estados Diferentes**

Um dos grandes benefícios da POO é a capacidade de criar milhares de objetos únicos a partir do mesmo código-fonte.

```python
# Instanciando dois carros diferentes
carro_a = Veiculo("Ford", "Mustang")
carro_b = Veiculo("Fiat", "Uno")

carro_a.ligar()

print(f"Carro A ({carro_a.modelo}) está ligado? {carro_a.ligado}")
print(f"Carro B ({carro_b.modelo}) está ligado? {carro_b.ligado}")
```

####**O que observar:**
<a>Embora usem a mesma classe, `carro_a` e `carro_b` são independentes. Ligar um não afeta o outro. Esta modularidade é o que permite, por exemplo, gerenciar múltiplos usuários ou produtos em um sistema de ADS sem confusão de dados.</a>
