### **1. Estruturas de Dados: Series e DataFrames**

Nesta etapa, criaremos as estruturas base do Pandas. Uma Series funciona como uma coluna rotulada, enquanto o DataFrame representa uma tabela completa. Utilizaremos dicionários Python para carregar esses dados manualmente, simulando a organização inicial descrita na Seção 4.1.

```python
import pandas as pd

# 1. Criando uma Series (Dados Unidimensionais)
idades = pd.Series([18, 20, 19, 24], name="Idade")

# 2. Criando um DataFrame (Dados Bidimensionais)
dados_alunos = {
    'Nome': ['Murilo', 'Rebeca', 'Gabriel', 'Laura'],
    'Curso': ['ADS', 'Medicina', 'ADS', 'Direito'],
    'Nota': [9.5, 8.0, 7.5, 6.0]
}

df_alunos = pd.DataFrame(dados_alunos)

print("--- Estrutura Series ---")
print(idades)
print("\n--- Estrutura DataFrame ---")
print(df_alunos)
```

#### **O que observar:**
<a>O experimento demonstra a diferença visual e estrutural entre as duas classes do Pandas. O DataFrame organiza os dados em linhas e colunas com índices, facilitando o acesso. Conforme Scheffer destaca na Seção 4.1, essa estrutura permite que operemos em colunas inteiras de forma vetorizada, o que é muito mais eficiente do que usar laços de repetição tradicionais.</a>

---
### **2. Leitura e Análise Exploratória de Dados**

Evoluindo para a Seção 4.2, simularemos a leitura de um arquivo estruturado. Utilizaremos métodos do Pandas para extrair informações rápidas da tabela, como os primeiros registros, tipos de dados e estatísticas descritivas (média, mínimo e máximo).

```python
# Simulando a análise de um conjunto de dados
print("[*] Iniciando Análise Exploratória...")

# Exibindo informações gerais sobre o DataFrame criado anteriormente
print("\nInformações do DataFrame:")
print(df_alunos.info())

print("\nEstatísticas Descritivas (Coluna Nota):")
print(df_alunos['Nota'].describe())

media_notas = df_alunos['Nota'].mean()
print(f"\nA média geral de notas é: {media_notas:.2f}")
```

#### **O que observar:**
<a>Neste teste, utilizamos o método `describe()`, que automatiza cálculos estatísticos complexos. **Nota Técnica:** No ambiente real, utilizaríamos `pd.read_csv()` ou `pd.read_excel()` para carregar milhares de linhas. A capacidade do Pandas de resumir dados instantaneamente valida a teoria da autora sobre a biblioteca ser o alicerce para análise prática de dados do mundo real.</a>

---
### **3. Filtros e Exportação de Dados**

Para concluir os conceitos da Unidade 4, aplicaremos filtros lógicos para selecionar apenas subconjuntos dos dados (ex: apenas alunos de ADS) e exportaremos o resultado para um novo arquivo CSV, garantindo a persistência da manipulação realizada.

```python
import matplotlib.pyplot as plt

# 1. Aplicando Filtro: Alunos apenas do curso de ADS
print("[*] Filtrando alunos de ADS...")
df_ads = df_alunos[df_alunos['Curso'] == 'ADS']

# 2. Visualização: Comparativo de Notas Médias por Curso
print("\n[*] Gerando gráfico comparativo...")
medias_por_curso = df_alunos.groupby('Curso')['Nota'].mean()

plt.figure(figsize=(8, 5))
medias_por_curso.plot(kind='bar', color=['orange', 'green', 'blue'])
plt.title('Média de Notas por Curso')
plt.xlabel('Cursos')
plt.ylabel('Nota Média')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# 3. Exportando os dados filtrados para um arquivo CSV
nome_arquivo = "alunos_ads_filtrados.csv"
df_ads.to_csv(nome_arquivo, index=False)
print(f"\n[SUCESSO] O arquivo '{nome_arquivo}' foi gerado.")
```

#### **O que observar:**
<a>O experimento demonstra o ciclo completo da análise de dados: coleta, filtragem e visualização. O gráfico de barras permite identificar rapidamente tendências (como a média de desempenho por curso) que seriam difíceis de notar apenas olhando para a tabela bruta.  
>**Nota Técnica**: Utilizamos o método `groupby` do Pandas para agregar os dados antes da plotagem, uma técnica fundamental da Seção 4.2 que transforma registros individuais em inteligência de negócio visual.</a>
